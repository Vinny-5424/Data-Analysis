```python
space = " "
ast ="*"
max = 5
min = 1

while max>=1:
    print(space * max + ast * min)
    max-=1
    min+=1

while min>0:
    print(space * max + ast * min)
    max+=1
    min-=1
```

```python
#Constructing a diamond
space = " "
ast ="*"
max = 5
min = 1

while max>=1:
    print(space * max + ast * min + space * max)
    max-=1
    min+=2

while min > 0:
    print(space * max + ast * min + space * max)
    min-=2
    max+=1
```

```python
#prediction of plant yield based on temp, rainfall, and humidity
kano = [73, 63, 43]
kaduna = [91, 88, 64]
jos = [87, 134, 58]
ondo = [102, 43, 37]
enugu= [69, 96, 70]
weights = [0.3, 0.2, 0.5]


def crop_yield(region, weights):
    result = 0
    #a zip is use to pair values of same index in two lists
    for a, b in zip(region, weights):
        result+=a*b
    return result

crop_yield(kano,weights)
```
