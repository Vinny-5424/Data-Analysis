# Simple project - Nut butter sales

```python
np.random.seed(4)
#Number of jars sold
sales_amount = np.random.randint(20, size=(5,3))
sales_amount
```

```python
import pandas as pd
#create Weekly sales
weekly_sales = pd.DataFrame(sales_amount, index=["Mon", "Tue", "Wed", "Thu", "Fri"], columns=["Almond butter", "Peanut butter", "Cashew butter"])
weekly_sales
```

```python
#Create price array
prices = np.array([10, 8, 12])
prices
```

```python
#Create butter_prices Dataframe
butter_prices = pd.DataFrame(prices.reshape(1,3), index=["Price"], columns=["Almond butter", "Peanut butter", "Cashew butter"])
butter_prices
```

```python
total_sales = np.matmul(sales_amount, prices)
total_sales
```

```python
weekly_sales["Total price"] = total_sales
weekly_sales
```
